## 2. Series (Tek Boyutlu Veri Yapısı)

`Series`, pandas'ın en temel veri yapısıdır. Tek bir sütundan oluşan, **indeksli** bir dizidir.

Aşağıda, veri setimizdeki `weekly_rank` (haftalık sıralama) sütununu tek başına bir Series olarak ele alacağız (veriyi az sonra tam olarak okuyacağız; burada kavramı göstermek için birkaç örnek değerle basit bir Series oluşturuyoruz).

```python
# Örnek haftalık sıralamalardan basit bir Series oluşturma
weekly_ranks_sample = pd.Series([1, 3, 5, 2, 7], name='weekly_rank')
print(weekly_ranks_sample)
print("\nVeri tipi:", type(weekly_ranks_sample))
print("Ortalama haftalık sıralama:", weekly_ranks_sample.mean())
```

## 5. Veri Seçme ve Filtreleme

- **Sütun seçme**: `df['sutun']` veya `df[['sutun1','sutun2']]`
- **`.loc[]`**: Etiket bazlı seçim
- **`.iloc[]`**: Konum bazlı seçim
- **Koşullu filtreleme**: `df[df['sutun'] == deger]`

```python
# Sadece belirli sütunları seçme
df[['show_title', 'category', 'season_title']].head()
```

```python
# .loc ile ilk satırın belirli sütunlarını seçme
df.loc[0, ['show_title', 'category', 'season_title', 'country_name']]
```

```python
# Koşullu filtreleme: 'Films' (Filmler) kategorisindeki yayınlar
df[df['category'] == 'Films'].head()
```

```python
# Birden fazla koşul: 'Filmler' kategorisindeki ve haftalık sıralaması 3 veya daha düşük olanlar
df[(df['category'] == 'Films') & (df['weekly_rank'] <= 3)].head()
```

## 6. Veri Düzenleme

- **Yeni sütun ekleme**: Mevcut sütunlardan türetme (örn. indirim yüzdesi)
- **`.apply()`**: Her satıra özel bir fonksiyon uygular
- **Sütun silme / yeniden adlandırma**

```python
# Performans puanı hesaplayan yeni bir sütun ekleme
# (Haftalık sıralama ne kadar düşükse ve Top 10'da ne kadar uzun kaldıysa o kadar iyi bir skor)
df['performance_score'] = (11 - df['weekly_rank']) * df['cumulative_weeks_in_top_10']
df[['show_title', 'weekly_rank', 'cumulative_weeks_in_top_10', 'performance_score']].head()
```

```python
# apply() ile kategorik bir sütun türetme: yüksek/düşük performans etiketi
# Örneğin, 'performance_score' 20'den yüksekse 'Yüksek', değilse 'Düşük' olarak etiketleyelim.
df['performance_category'] = df['performance_score'].apply(lambda x: 'Yüksek' if x > 20 else 'Düşük')
df[['show_title', 'performance_score', 'performance_category']].head()
```

```python
# Sütun ismini değiştirme (Türkçeleştirme örneği)
# Netflix verisine uygun sütunları yeniden adlandıralım
df_tr = df.rename(columns={'show_title': 'dizi_film_adi', 'country_name': 'ulke_adi'})
df_tr[['dizi_film_adi', 'ulke_adi']].head()
```

## 7. Sıralama ve Gruplama

- **Sıralama**: `df.sort_values('sutun', ascending=False)`
- **Gruplama (`groupby`)**: Kategorik bir sütuna göre grupları oluşturup toplama işlemi yapar
- **`.agg()`**: Aynı anda birden fazla istatistik hesaplar

```python
# Performans puanına göre sıralama
# En yüksek performans skoruna sahip dizi/filmleri gösterelim
df.sort_values('performance_score', ascending=False)[['show_title', 'performance_score', 'cumulative_weeks_in_top_10', 'weekly_rank']].head()
```

```python
# Ülkelere göre gruplayıp performans skorunun toplamını, kayıt sayısını ve ortalama değerini hesaplama
df.groupby('country_name').agg(
    toplam_performance_skoru=('performance_score', 'sum'),
    performance_skoru_kayit_sayisi=('performance_score', 'count'),
    ortalama_performance_skoru=('performance_score', 'mean')
).sort_values('ortalama_performance_skoru', ascending=False)
```

## 8. Birleştirme İşlemleri (Concat & Merge)

- **`pd.concat([df1, df2])`**: DataFrame'leri alt alta birleştirir.
- **`pd.merge(df1, df2, on='ortak_sutun', how='...')`**: SQL'deki JOIN işlemine benzer.


```python
# concat örneği: iki alt kümeyi alt alta birleştirme
df_films = df[df['category'] == 'Films'].head(3)
df_tv = df[df['category'] == 'TV'].head(3)

df_birlesik = pd.concat([df_films, df_tv])
df_birlesik[['show_title', 'category', 'weekly_rank']]
```

```python
# merge örneği: ülkelere yapay bölge bilgisi ekleme
# Netflix veri setinde 'store_name' bulunmadığından, 'country_name' sütununu kullanalım.
country_regions = pd.DataFrame({
    'country_name': df['country_name'].dropna().unique()
})
# Örnek bölge ataması (gerçek veride bulunmadığı için gösterim amaçlı ekliyoruz)
regions = ['Kuzey Amerika', 'Avrupa', 'Asya', 'Güney Amerika', 'Afrika', 'Okyanusya', 'Orta Doğu']
num_countries = len(country_regions)
country_regions['region'] = [regions[i % len(regions)] for i in range(num_countries)]

df_regional = pd.merge(df, country_regions, on='country_name', how='left')
df_regional[['country_name', 'region', 'performance_score']].head()

```

## 9. Uygulamalı Örnek: Ülke Bazında Performans Skoru

Öğrenilen tüm kavramları birleştirip gerçek veri üzerinde küçük bir analiz yapalım:

```python
# Ülke bazında toplam performans skoru ve kayıt sayısı
ulke_ozet = df.groupby('country_name').agg(
    toplam_performance_skoru=('performance_score', 'sum'),
    kayit_sayisi=('performance_score', 'count')
).sort_values('toplam_performance_skoru', ascending=False)

print("Ülke bazında özet:")
print(ulke_ozet)

en_iyi_ulke = ulke_ozet['toplam_performance_skoru'].idxmax()
print(f"\nEn yüksek toplam performansa sahip ülke: {en_iyi_ulke}")
```

```python
# En yüksek performansa sahip dizi/filmler (toplam performans skoru bazında)
en_iyi_performansli_dizi_film = df.groupby('show_title')['performance_score'].sum().sort_values(ascending=False)
print("En yüksek performansa sahip ilk 5 dizi/film:")
print(en_iyi_performansli_dizi_film.head())
```
